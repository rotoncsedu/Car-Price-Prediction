# <h2 style= 'text-align:center; color:#27AE60; font-weight:bold;font-size:40px;  text-shadow:0 0 3px rgba(15,52,96,0.6);'>Car Age Analysis!</h2>

```python
display(Markdown("### <span style='color:#27AE60; font-weight:bold;font-size:24px; text-shadow:0 0 3px rgba(15,52,96,0.6);'>Creating Car_Age Feature</span>"))

# Car_Age = number of years since the car was manufactured, relative to the most recent model year present in the dataset
data['Car_Age'] = data['Year'].max() + 1 - data['Year']

display(data[['Year', 'Car_Age']].head())
```

```python
display(Markdown("### <span style='color:#27AE60; font-weight:bold;font-size:24px; text-shadow:0 0 3px rgba(15,52,96,0.6);'>Selling_Price vs Car_Age</span>"))

fig, axes = plt.subplots(1, 2, figsize=(16, 5))

sns.scatterplot(data=data, x='Car_Age', y='Selling_Price', alpha=0.5, ax=axes[0])
axes[0].set_title('Selling_Price vs Car_Age')

avg_price_by_age = data.groupby('Car_Age')['Selling_Price'].mean().reset_index()
sns.lineplot(data=avg_price_by_age, x='Car_Age', y='Selling_Price', marker='o', ax=axes[1])
axes[1].set_title('Average Selling_Price by Car_Age')

plt.tight_layout()
plt.show()

display(Markdown(f"**Correlation between Car_Age and Selling_Price:** {data['Car_Age'].corr(data['Selling_Price']):.3f}"))
```

# <h2 style= 'text-align:center; color:#27AE60; font-weight:bold;font-size:40px;  text-shadow:0 0 3px rgba(15,52,96,0.6);'>Key Findings!</h2>

### Observations

- **Target distribution:** `Selling_Price` is right-skewed (see skewness value above) — most cars sell in the lower price range, with a smaller number of high-priced cars pulling the tail to the right. This suggests a log-transform of `Selling_Price` may help linear models.
- **Most influential features:** `Present_Price` shows the strongest positive correlation with `Selling_Price` — the current market price of the car is the single best predictor of its resale price. `Car_Age` shows a clear negative relationship: older cars sell for less, confirming depreciation over time.
- **Moderate/weak signal:** `Kms_Driven` has a much weaker relationship with `Selling_Price` than expected, and `Owner` shows very little variation/signal since the vast majority of cars have had 0 previous owners.
- **Categorical effects:** Diesel cars and automatic-transmission cars tend to have a higher median `Selling_Price` than petrol/CNG and manual cars respectively, based on the box plots above. `Seller_Type` (Dealer vs Individual) also shows a visible price gap.
- **Redundant / low-signal features:** `Car_Name` has very high cardinality (many unique values) and duplicates information already captured indirectly through `Present_Price`/brand-level pricing — it is a candidate to drop or to bucket into a "Brand" feature. `Year` becomes redundant once `Car_Age` is engineered, since they carry the same information.

### Planned Preprocessing

- Drop or bucket `Car_Name` into a lower-cardinality `Brand` feature; drop the raw `Year` column once `Car_Age` is created.
- One-hot / label-encode categorical columns (`Fuel_Type`, `Seller_Type`, `Transmission`).
- Consider a log-transform on `Selling_Price` (and possibly `Present_Price`, `Kms_Driven`) to reduce right-skew before modeling.
- Scale numerical features (`Present_Price`, `Kms_Driven`, `Car_Age`) if using distance-based or regularized models.
- No missing values were found, so no imputation is required; duplicates were already removed earlier in this notebook.
